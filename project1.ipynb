## The 'easy' way (using Pandas)

I decided to take a closer look at the variable 'RIDAGEYR' to see how the age of the survey respondents changed due to the effects of the pandemic. We can calculate the mean, median, and mode of this variable using the pandas module:

```python
# Calculating the mean:

mean_age = nhanes["RIDAGEYR"].mean()

# Calculating the median:

median_age = nhanes["RIDAGEYR"].median()

# Calculating the mode:

mode_age = nhanes["RIDAGEYR"].mode()
```

As seen from the outputs above, we see that, between the months of August 2021 and August 2023 for the CDC's NHANES survey, the **mean age of respondents was 38.32 years**, the **median age of respondents was 37 years**, and the **mode age for respondents was 80 years**.

# The 'hard' way (not using Pandas)

Now we can take a more difficult approach to do the same analysis without making use of the pandas module, meaning utilizing base Python. In this case, the main difficulty is that base Python does not have a way to read SAS files, so I had to first convert to a CSV file using a simple command that does use pandas solely for this purpose (approval by Prof. Feldman was granted).

To do this I checked StackOverflow and found a useful example here: https://stackoverflow.com/questions/77951378/conversion-of-xpt-to-csv

```python
# Converting the NHANES SAS file in to CSV:

nhanes.to_csv("DEMO_L.csv", index=False)
```

Then, as hinted in the assignment instructions, I made use of the file-reading method covered during the first half of our CIC course to read each rows for the 'RIDAGEYR' column and then storing each of the ages into a dictionary:

**Disclaimer:** For this part, I asked Google Gemini "when using the readline function in python, how can I select an specific column from the csv I am reading?". I saw from here that I needed to first split the line read by commas (which makes sense because it is a CSV), and then access the desired column by its zero-based index. This guided my approach below.

```python
# Opening my converted CSV file

nhanes_csv = open("DEMO_L.csv", "r")

# Now I can use the 'readline()' function to read the first line, which corresponds to the headers.
# We also need to remove the new line symbol like we did in previous lecture:

header_line = nhanes_csv.readline()
header_line = header_line.rstrip("\n")

# Now, as mentioned above, I need to split the column line by commas
# in order to access the individual column names:

col_names = header_line.split(",")

# Now I am getting the zero-based index of the column I want to analyze,
# which in this case corresponds to age:

agecol_index = col_names.index("RIDAGEYR")

# Creating my empty dictionary to store each of the age values in my data set:

nhanes_ages = []

# Now I can keep reading the rest of the file:

next_line = nhanes_csv.readline()

# Now we can use a similar strategy as the one used during Lecture 12 when we used
# a 'while loop' to read one line at a time:

while next_line:
    # We need to remove the new line character again like above:

    next_line = next_line.rstrip("\n")

    # We need to also split the observations by commas just like above.
    # I will make use of the variable referring to the column for ages:

    split_next_line = next_line.split(",")

    # Now, I need to read the row value for the age column:

    age_value = split_next_line[agecol_index]

    # Appending the individual age values to the list I created before and
    # making sure that it is a float:

    nhanes_ages.append(float(age_value))

    # Now reading the next line to keep the loop going:

    next_line = nhanes_csv.readline()

# Closing my file:

nhanes_csv.close()
```

### Calculating the Mean

Now that I created a list with the age values from the NHANES data set....

```python
# We can create a 'bucket' counter to make
#

total_age = 0


for age_value in nhanes_ages:
    total_age = (total_age) + (age_value)

# Now we can do the mean calculation by hand:

mean_nhanes_age = (total_age) / (len(nhanes_ages))


# Now we can print the value:

print(mean_nhanes_age)
```

### Calculating the Median

Now that I created a list with the age values from the NHANES data set....

```python
# We first need to sort our list to make sure the
# age values are in order:

nhanes_ages.sort()

# Obtaining the number of observations:

obs = len(nhanes_ages)
```

Notice that for the median we have two possible scenarios: the number of observations is odd or the number of observations is even...

```python
# We can create an 'if' statement for where we first assess for the
# 'even number of observations' case. Here we can use the indexing
# brackets in order to obtain the median value:

if obs % 2 == 0:
    index_value = obs // 2
    median_nhanes_age = nhanes_ages[index_value]

else:
    index_value1 = obs // 2
    value1 = nhanes_ages[index_value1]

    index_value2 = (obs // 2) + 1
    value2 = nhanes_ages[index_value1]

    median_nhanes_age = (value1 + value2) / 2

# We can now print the median value:

print(median_nhanes_age)
```

## Calculating the Mode

Next, to calculate the mode, we need to create a counter for each of the potential age values, and then see which of the age values has the most entries. For this, I am using a dictionary per the hint provided by Prof. Feldman in the instructions:

```python
# We can first and foremost create an empty dictionary in order to
# keep track of the counts:

age_value_counter = {}

# Now, we can create a for loop where we go through each of the age
# values and then create an if statement that let's us add to the counter
# for each age value:

for age_value in nhanes_ages:
    if age_value in age_value_counter:
        age_value_counter[age_value] = age_value_counter[age_value] + 1

    else:
        age_value_counter[age_value] = 1
```

Now that we have a dictionary of all of our counts, we can go ahead and create another loop to look for the highest_count:

```python
# Creating my bucket variables so I can call
# them later:

most_obs = 0

mode_nhanes_age = 0

for age_value in age_value_counter:
    if age_value_counter[age_value] > most_obs:
        most_obs = age_value_counter[age_value]

        mode_nhanes_age = age_value


print(mode_nhanes_age)
```

# Data Visualization

We can create a similar visualization compared to the one presented in the Project 1 example by first and foremost creating a dictionary with all the possible ranges of ages contained in the NHANES data set:

```python
# Creating my dictionary from which I will do counters
# pertaining to each of the age groups:


age_groups = {
    "0-10 years": 0,
    "11-20 years": 0,
    "21-30 years": 0,
    "31-40 years": 0,
    "41-50 years": 0,
    "51-60 years": 0,
    "61-70 years": 0,
    "71-80+ years": 0,
}

# Then I can create a for loop to start adding people to the dictionary counter system:

for age_value in nhanes_ages:
    if (age_value >= 0) and (age_value < 11):
        age_groups["0-10 years"] = age_groups["0-10 years"] + 1

    elif (age_value >= 11) and (age_value < 21):
        age_groups["11-20 years"] = age_groups["11-20 years"] + 1

    elif (age_value >= 21) and (age_value < 30):
        age_groups["21-30 years"] = age_groups["21-30 years"] + 1

    elif (age_value >= 31) and (age_value < 40):
        age_groups["31-40 years"] = age_groups["31-40 years"] + 1

    elif (age_value >= 41) and (age_value < 50):
        age_groups["41-50 years"] = age_groups["41-50 years"] + 1

    elif (age_value >= 51) and (age_value < 60):
        age_groups["51-60 years"] = age_groups["51-60 years"] + 1

    elif (age_value >= 61) and (age_value < 70):
        age_groups["61-70 years"] = age_groups["61-70 years"] + 1

    elif age_value >= 71:
        age_groups["71-80+ years"] = age_groups["71-80+ years"] + 1


print("Age Distribution of NHANES Participants (2021-2023)")

print("Each star (*) represents 100 observations")

for age_group in age_groups:
    stars_to_print = age_groups[age_group] // 100

    star_bar = stars_to_print * "*"

    print(age_group, ": ", star_bar)
```
